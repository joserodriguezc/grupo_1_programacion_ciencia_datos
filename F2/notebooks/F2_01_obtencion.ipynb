## 03 · Extraer diputados

Primero se busca el periodo correspondiente a cada votación y luego se integra en una variable para la consulta `retornarDiputadosXPeriodo` para el período legislativo vigente

```python
def buscar_periodo(fecha_votacion, df_periodos):
    inicios = pd.to_datetime(df_periodos["fecha_inicio"])
    terminos = pd.to_datetime(df_periodos["fecha_termino"])
    vigente = df_periodos[(inicios <= fecha_votacion) & (fecha_votacion <= terminos)]
    if vigente.empty:
        return None
    return vigente.iloc[0]["periodo_id"]

fechas_votacion = pd.to_datetime(df_votaciones["Fecha"])
df_votaciones["periodo_id"] = fechas_votacion.apply(lambda f: buscar_periodo(f, df_periodos))

df_votaciones[["Id", "Fecha", "periodo_id"]]
```

```python
PERIODO_ID = df_votaciones["periodo_id"].iloc[0]
print(PERIODO_ID)
```

```python
extraer_diputados(PERIODO_ID)
```
